```python
from IPython import get_ipython
from IPython.display import display
# %%
import numpy as np 
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
 #length ke bahar na jaye graph
%matplotlib inline 
print("ready!")
# %%
from io import StringIO
data=StringIO('''
Roll_no,Branch,Gender,Study_Hours,Attendance,Maths,Python,ML
1,ECE,Female,3.5,70,77,55,30
2,CSE,Female,4.5,80,98,41,20
3,MCA,Male,5.9,45,35,97,98
4,BCA,Male,9.7,97,10,29,76
5,CE,Male,2.5,45,61,32,54
6,CSE,Female,3.2,45,87,63,23
''')
df=pd.read_csv(data)
df["Total"]=df[["Maths","Python","ML"]].sum(axis=1)
df["Percentage"]=(df["Total"] / 3).round(1)
df ["Grade"]="D"
df.loc[df["Percentage"]>=60,"Grade"]="C"
df.loc[df["Percentage"]>=70,"Grade"]="B"
df.loc[df["Percentage"]>=80,"Grade"]="A"
df.head()

print(df)
# %%
#Draw a bar chart showing the no of student in each Branch(CSE,IT,ECE)
counts = df["Branch"].value_counts()

plt.bar(counts.index, counts.values,color=["red","yellow","brown"])

plt.xlabel("Branch")
plt.ylabel("Number of Students")
plt.title("Number of Students in Each Branch")

plt.show()

# %%
#Line plot of the plot of the python marks of the first 5 student(x=Roll_no).use a red dashed line with circle markers

plt.plot(df["Roll_no"].head(5),
         df["Python"].head(5),
         color="violet",
         linestyle='--',
         marker="o")
plt.xticks([1.5, 2.1,3.1,4.1], [1, 2,3,4])

plt.xlabel("Roll Number")
plt.ylabel("Python Marks")
plt.title("Python Marks of First 5 Students")

plt.show()
# %%
# two lines with legend 
plt.plot(df["Roll_no"].head(7), df["Python"].head(7),
         color='red', marker='o', linestyle='--', label='Python')

plt.plot(df["Roll_no"].head(7), df["Maths"].head(7),
         color='blue', marker='s', linestyle='-', label='Maths')

plt.xlabel("Roll Number")
plt.ylabel("Marks")
plt.title("Python vs Maths Marks")

plt.legend()
plt.show()
# %%
#Draw a histogram of attendence with 10bins,orangeand white edges 
plt.hist(df["Attendance"], bins=10, color='orange', edgecolor='white')

plt.xlabel("Attendance")
plt.ylabel("Frequency")
plt.title("Distribution of Student Attendance")

plt.show()
# %%
#subplot(2*2) and save


fig, ax = plt.subplots(2, 2, figsize=(10, 8))

# 1. Bar chart
counts = df["Branch"].value_counts()
ax[0, 0].bar(counts.index, counts.values)
ax[0, 0].set_title("Students in Each Branch")

# 2. Line plot
ax[0, 1].plot(df["Roll_no"].head(5), df["Python"].head(5),
              color='red', marker='o')
ax[0, 1].set_title("Python Marks")

# 3. Histogram
ax[1, 0].hist(df["Attendance"], bins=10,
              color='orange', edgecolor='white')
ax[1, 0].set_title("Attendance Distribution")

# 4. Scatter plot
ax[1, 1].scatter(df["Python"], df["Maths"])
ax[1, 1].set_title("Python vs Maths")

# Adjust layout
plt.tight_layout()

# Save figure
plt.savefig("subplot.png")

# Display figure
plt.show()
```
