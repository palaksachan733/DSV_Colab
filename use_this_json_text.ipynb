```python
from IPython import get_ipython
from IPython.display import display
# %%
t=('{"shop":"Fresh Mart","open":true,"Owner":null,"items":[''{"name":"Rice","price":60,"qty":10},''{"name":"oil","price":150,"qty":4},''{"name":"tea","price":120,"qty":6}]}')
# %%
#load t into a python object and print its type.
import json

data = json.loads(t)

print(data)
print(type(data))
# %%
# print the name of the second item and the number of items

print(data["items"][1]["name"])

print(len(data["items"]))
# %%
# print the value of the owner what python values is it 
print(data["Owner"])
print(type(data["Owner"]))
# %%
#convert the object back to json text with indentation 2
json_text = json.dumps(data, indent=2)

print(json_text)
```
