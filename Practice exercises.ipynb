```python
from IPython import get_ipython
from IPython.display import display
# %%
import pandas as pd;
# %%
# create a csv with no header and 5 columns.load it with your own column names and makes the first columns the index
data = [
    [101, "Palak", 20, "CSE", 85],
    [102, "Aman", 21, "ECE", 90],
    [103, "Riya", 20, "CSE", 88],
    [104, "Rahul", 22, "IT", 78]
]

df = pd.DataFrame(data)
df.to_csv("students.csv", index=False, header=False)

df = pd.read_csv(
    "students.csv",
    header=None,
    names=["Roll_No", "Name", "Age", "Branch", "Marks"],
    index_col=0
)

print(df)
# %%
# write a file using ; as the delimiter and-999 for mising values.read it so -999 becomes NaN
data = {
    "Name": ["Palak", "Aman", "Riya"],
    "Age": [20, -999, 21],
    "Marks": [85, 90, -999]
}

df = pd.DataFrame(data)


df.to_csv("students.csv", sep=";", index=False)


df = pd.read_csv("students.csv", sep=";", na_values=-999)

print(df)
# %%
# READ A BIG CSV IN CHUNKS OF 2000 ROWS AND COMPUTE THE MEAN OF ONE COLUMNS PER KEY ACROSS ALL CHUNKS
# %%
# EXPORTS  A DATAFRAME WITH ORIENTS ="RECORDS",READ IT BACK WITH PD.READ_JSON,AND CHECK EQUALITY WITH .EQUALS
df = pd.DataFrame({
    "Name": ["Palak", "Aman", "Riya"],
    "Age": [20, 21, 19],
    "Marks": [85, 90, 88]
})


df.to_json("students.json", orient="records")


df2 = pd.read_json("students.json", orient="records")

print(df.equals(df2))
# %%
#WRITE A SHORT HTML TABLE BY HAND AND LOAD IT WITH READ_html
import pandas as pd
from io import StringIO


html = """
<table>
    <tr>
        <th>Name</th>
        <th>Age</th>
        <th>Marks</th>
    </tr>
    <tr>
        <td>Palak</td>
        <td>20</td>
        <td>85</td>
    </tr>
    <tr>
        <td>Aman</td>
        <td>21</td>
        <td>90</td>
    </tr>
</table>
"""
df = pd.read_html(StringIO(html))[0]

print(df)
```
